**Generate ANDNOT function using McCulloch-Pitts neural net by a python program.**

```python
import numpy as np


def McCulloch(dot, T):
    return 1 if dot >= T else 0

inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
weights = ([1, -1])
T = 1

dots = inputs @ weights

for i in range(4):
    out = McCulloch(dots[i], T)
    print(f"{inputs[i]}   |    {out}")
```
