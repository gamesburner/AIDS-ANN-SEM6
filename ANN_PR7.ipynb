**Write a python program to show Back Propagation Network for XOR function with Binary Input and Output**

```python
import numpy as np

sig  = lambda x: 1 / (1 + np.exp(-x))
dSig = lambda x: x * (1 - x)

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

np.random.seed(1)
W1 = np.random.uniform(-1, 1, (2, 2))  
W2 = np.random.uniform(-1, 1, (2, 1)) 
b1 = np.zeros((1, 2))   
b2 = np.zeros((1, 1))

# Train
for _ in range(10000):
    # Forward
    h = sig(X @ W1 + b1)
    o = sig(h @ W2 + b2)      

    # Backward
    e2 = (y - o) * dSig(o)         
    e1 = (e2 @ W2.T) * dSig(h)     

    # Update
    W2 += h.T @ e2 * 0.1
    W1 += X.T @ e1 * 0.1
    b2 += e2.sum(axis=0) * 0.1  
    b1 += e1.sum(axis=0) * 0.1 

# Results
for i in range(4):
    print(f"Input: {X[i]} → Output: {round(o[i][0])}")
```
