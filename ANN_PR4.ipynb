**With a suitable example demonstrate the perceptron learning law with its decision regions using python. Give the output in graphical form.**

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[2,3],[1,1],[2,1],[3,1.5],[4,2],[3,3.5],[5,3],[6,2.5]])
y = np.array([1,-1,-1,-1,1,1,1,1])

w, b = np.zeros(2), 0
for _ in range(10):
    for xi, t in zip(X, y):
        e = t - np.sign(np.dot(w, xi) + b)
        w += 0.1 * e * xi
        b += 0.1 * e

print("w:", w, "b:", b)

x_vals = np.linspace(0, 7, 200)
y_vals = -(w[0] * x_vals + b) / w[1]

plt.scatter(X[y==1, 0],  X[y==1, 1],  color='blue', marker='o', label='+1')
plt.scatter(X[y==-1, 0], X[y==-1, 1], color='red',  marker='x', label='-1')
plt.plot(x_vals, y_vals, 'k--', label='Boundary')
plt.legend()
plt.title("Perceptron")
plt.show()
```
